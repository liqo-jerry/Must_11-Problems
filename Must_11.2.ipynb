## Problem 2 — Find Larger and Smaller of Two Numbers

### 📌 Question

Write a Python program to input two numbers and display the larger and smaller number.

### 💡 Logic

Suppose the numbers are a and b.

If a > b, then a is larger and b is smaller.
Otherwise, b is larger and a is smaller.

### 🐍 Python Program

```python
# Program to find the larger and smaller of two numbers

# Step 1: Input two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

# Step 2: Compare the two numbers
if a > b:
    print("Larger number =", a)
    print("Smaller number =", b)
elif b > a:
    print("Larger number =", b)
    print("Smaller number =", a)
else:
    print("Both numbers are equal")
```

### 📖 Important Concept

if, elif, and else are conditional statements.

They allow Python to make decisions.
