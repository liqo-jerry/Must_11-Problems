## Problem 9 — GCD and LCM of Two Integers

### 📌 Question

Compute the Greatest Common Divisor and Least Common Multiple of two integers.

💡 Example

For:

12 and 18

GCD = 6

LCM = 36

### 🐍 Python Program

```python
# Program to find GCD and LCM of two numbers

# Input two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

# Store the original values
x = a
y = b

# Find GCD
while y != 0:

    remainder = x % y

    x = y
    y = remainder

# x now contains the GCD
gcd = x

# Find LCM
lcm = (a * b) // gcd

# Display results
print("GCD =", gcd)
print("LCM =", lcm)
```

### 📖 Important

The expression:

a * b

multiplies the numbers.

The expression:

//

performs integer division.
