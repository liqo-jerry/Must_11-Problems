## Problem 7 — Check Prime or Composite

### 📌 Question

Input a number and check whether it is a prime or composite number.

### 💡 Logic

- A prime number has exactly two factors: 1 and itself

Examples: 2, 3, 5, 7, 11

- A composite number has more than two factors.

### 🐍 Python Program

```python
# Program to check whether a number is Prime or Composite

# Input the number
num = int(input("Enter a number: "))

# Special case
if num <= 1:
    print("The number is neither Prime nor Composite")

else:
    # Assume the number is prime
    prime = True

    # Check divisibility
    for i in range(2, num):

        if num % i == 0:
            prime = False
            break

    # Display result
    if prime:
        print(num, "is a Prime Number")
    else:
        print(num, "is a Composite Number")
```

### 📖 Why use break?

When we find one factor, we already know the number is composite.

So there is no need to continue the loop.
