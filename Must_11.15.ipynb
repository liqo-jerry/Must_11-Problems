## Problem 15 — Dictionary of Students and Marks

### 📌 Question

Create a dictionary containing the roll number, name and marks of n students and display the names of students who have scored above 75.

### 💡 Structure

We can create a dictionary like:

Roll Number → Name and Marks

For example:

students = {
    101: ["Rahul", 80],
    102: ["Aman", 65],
    103: ["Priya", 90]
}

### 🐍 Python Program

```python
# Program to create a dictionary of students
# and display students having marks above 75

# Input the number of students
n = int(input("Enter the number of students: "))

# Create an empty dictionary
students = {}

# Input student details
for i in range(n):

    # Input roll number
    roll = int(input("Enter roll number: "))

    # Input student name
    name = input("Enter student name: ")

    # Input marks
    marks = float(input("Enter marks: "))

    # Store name and marks in the dictionary
    students[roll] = [name, marks]

# Display students with marks above 75
print("\nStudents who scored above 75:")

for roll in students:

    # Get the student's name
    name = students[roll][0]

    # Get the student's marks
    marks = students[roll][1]

    # Check marks
    if marks > 75:
        print(name)
```
