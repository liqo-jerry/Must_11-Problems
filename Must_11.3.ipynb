## Problem 3 — Find Largest and Smallest of Three Numbers

### 📌 Question

Write a Python program to input three numbers and display the largest and smallest number.

### 🐍 Python Program

```python
# Program to find the largest and smallest of three numbers

# Step 1: Input three numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

# Step 2: Find the largest number
if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c

# Step 3: Find the smallest number
if a <= b and a <= c:
    smallest = a
elif b <= a and b <= c:
    smallest = b
else:
    smallest = c

# Step 4: Display the result
print("Largest number =", largest)
print("Smallest number =", smallest)
```

📖 Explanation

The operator: (and)

means both conditions must be true.

For example:

if a >= b and a >= c:

means:

Is a greater than or equal to b AND a greater than or equal to c?

If yes, a is the largest.
