## Problem 1 — Input and Display a Welcome Message

### 📌 Question

Write a Python program to input a welcome message from the user and display it.

### 💡 Logic

We need to:

1. Ask the user to enter a welcome message.
2. Store the message in a variable.
3. Display the message using print().

### 🐍 Python Program

```python
# Program to input and display a welcome message

# Step 1: Take a welcome message from the user
message = input("Enter your welcome message: ")

# Step 2: Display the message
print("Welcome Message:", message)
```

### 📖 Explanation

message = input("Enter your welcome message: ")

- input() is used to take data from the user.
- Whatever the user enters is stored in the variable message.

print("Welcome Message:", message)

- print() displays the message on the screen.
- The value stored in message is also displayed.
