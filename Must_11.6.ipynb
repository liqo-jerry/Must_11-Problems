## Problem 6 — Perfect Number, Armstrong Number and Palindrome

We need to check whether a number is:

- Perfect number
- Armstrong number
- Palindrome

### 6(a) Perfect Number

- A perfect number is a number whose proper divisors add up to the number itself.

Example:

6

Divisors: 1, 2, 3

1 + 2 + 3 = 6

Therefore, 6 is a perfect number.

### 🐍 Python Program

```python
# Program to check whether a number is a Perfect Number

# Input the number
num = int(input("Enter a number: "))

# Store the sum of divisors
sum = 0

# Check divisors from 1 to num-1
for i in range(1, num):

    if num % i == 0:
        sum = sum + i

# Check whether the sum equals the number
if sum == num:
    print(num, "is a Perfect Number")
else:
    print(num, "is not a Perfect Number")
```

### 6(b) Armstrong Number

For a 3-digit number, an Armstrong number is a number where:

$$ a^3+b^3+c^3 = number $$

Example:

153

1³ + 5³ + 3³
= 1 + 125 + 27
= 153

### 🐍 Python Program

```python
# Program to check whether a number is an Armstrong Number

# Input the number
num = int(input("Enter a number: "))

# Store the original number
original = num

# Store the sum
sum = 0

# Process each digit
while num > 0:

    # Get the last digit
    digit = num % 10

    # Add cube of the digit
    sum = sum + digit ** 3

    # Remove the last digit
    num = num // 10

# Compare the sum with original number
if sum == original:
    print(original, "is an Armstrong Number")
else:
    print(original, "is not an Armstrong Number")
```

### 6(c) Palindrome Number

A number is a palindrome if it remains the same when reversed.

Example:

121 → 121

### 🐍 Python Program

```python
# Program to check whether a number is a Palindrome

# Input the number
num = int(input("Enter a number: "))

# Store the original number
original = num

# Start reverse with 0
reverse = 0

# Reverse the number
while num > 0:

    # Get the last digit
    digit = num % 10

    # Add digit to reverse
    reverse = reverse * 10 + digit

    # Remove last digit
    num = num // 10

# Compare original and reverse
if original == reverse:
    print(original, "is a Palindrome")
else:
    print(original, "is not a Palindrome")
```
