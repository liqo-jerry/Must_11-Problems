## Problem 4 — Generate Patterns Using Nested Loops

### Pattern 1

### 💡 Logic

We need 5 rows.

Row 1 → 1 star

Row 2 → 2 stars

Row 3 → 3 stars

Row 4 → 4 stars

Row 5 → 5 stars

This requires a nested loop.

🐍 Python Program

```python
# Program to print Pattern-1

# Outer loop controls the rows
for i in range(1, 6):

    # Inner loop prints stars
    for j in range(1, i + 1):
        print("*", end="")

    # Move to the next line
    print()
```

### 📖 Understanding Nested Loops

There are two loops:

for i in range(1, 6):

This controls the rows.

for j in range(1, i + 1):

This controls the number of stars.

- end="" prevents Python from moving to the next line after every star.

### Pattern 2

### 🐍 Python Program

```python
# Program to print Pattern-2

# Outer loop controls the rows
for i in range(5, 0, -1):

    # Inner loop prints numbers
    for j in range(1, i + 1):
        print(j, end="")

    # Move to the next line
    print()
```

### 📖 Explanation

range(5, 0, -1)
means:
5
4
3
2
1

- So the number of digits decreases in every row.

### Pattern 3

### 🐍 Python Program

```python
# Program to print Pattern-3

# Store the alphabet
alphabet = "ABCDE"

# Outer loop controls the rows
for i in range(1, 6):

    # Inner loop prints characters
    for j in range(i):
        print(alphabet[j], end="")

    # Move to the next line
    print()
```

### 📖 Explanation

The string:

alphabet = "ABCDE"

has positions:

- A → 0
- B → 1
- C → 2
- D → 3
- E → 4

- Python uses index numbers starting from 0.
