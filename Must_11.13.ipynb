## Problem 13 — Swap Elements at Odd and Even Locations

### 📌 Question

Input a list of numbers and swap elements at the even location with elements at the odd location.

For example:

Original:
[10, 20, 30, 40, 50, 60]

After swapping:
[20, 10, 40, 30, 60, 50]

- Note: In Python, list positions/indexes start from 0.

Therefore:

Index:    0   1   2   3   4   5
Element: 10  20  30  40  50  60

### 🐍 Python Program

```python
# Program to swap elements at odd and even positions

# Input a list
numbers = [10, 20, 30, 40, 50, 60]

# Display original list
print("Original list:", numbers)

# Start from the first element
for i in range(0, len(numbers) - 1, 2):

    # Swap the elements
    numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]

# Display the changed list
print("List after swapping:", numbers)
```

### 📖 Important

- This:numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]
swaps two values.

- We can also write it using a temporary variable:

temp = numbers[i]

numbers[i] = numbers[i + 1]

numbers[i + 1] = temp

- The second version may be easier to understand initially.
