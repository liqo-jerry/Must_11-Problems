## Problem 11 — String Palindrome and Case Conversion

This problem has two parts:

Check whether a string is a palindrome.
Convert the case of characters.
11(a) Check String Palindrome

### 🐍 Python Program

```python
# Program to check whether a string is a palindrome

# Input the string
text = input("Enter a string: ")

# Reverse the string
reverse = text[::-1]

# Compare original and reverse
if text == reverse:
    print("The string is a Palindrome")
else:
    print("The string is not a Palindrome")
```

### 11(b) Convert Case

### 🐍 Python Program

```python
# Program to convert the case of characters

# Input the string
text = input("Enter a string: ")

# Convert to uppercase
print("Uppercase:", text.upper())

# Convert to lowercase
print("Lowercase:", text.lower())

# Swap uppercase and lowercase
print("Swapped case:", text.swapcase())
```
