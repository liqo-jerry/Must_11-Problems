## Problem 14 — Search for an Element in a List/Tuple

### 📌 Question

Input a list/tuple of elements and search for a given element.

### 🐍 Python Program

```python
# Program to search for an element in a list

# Create a list
numbers = [10, 20, 30, 40, 50]

# Input the element to search
search = int(input("Enter the number to search: "))

# Assume the element is not found
found = False

# Search each element
for num in numbers:

    if num == search:
        found = True
        break

# Display the result
if found:
    print("Element found in the list")
else:
    print("Element not found in the list")
```

### 📖 Main Concept

We compare every element with the value entered by the user:

if num == search:

If they are equal, the element has been found.
