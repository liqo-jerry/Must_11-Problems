## Problem 12 — Find Largest and Smallest Number in a List/Tuple

### 📌 Question

Find the largest and smallest number in a list or tuple.

### 🐍 Python Program Using List

```python
# Program to find the largest and smallest
# number in a list

# Input numbers
numbers = [25, 10, 45, 5, 30]

# Assume the first number is largest
largest = numbers[0]

# Assume the first number is smallest
smallest = numbers[0]

# Check every number
for num in numbers:

    # Check for largest
    if num > largest:
        largest = num

    # Check for smallest
    if num < smallest:
        smallest = num

# Display results
print("List =", numbers)
print("Largest number =", largest)
print("Smallest number =", smallest)
```

### 📖 Why start with numbers[0]?

We need an initial value for comparison.

The first element is a convenient starting point.
